## Dataset Configuration

```python
import pandas as pd
import glob  # since there are multiple files within each folder of datasets, we use glob to collectively use them

transaction_files = glob.glob("../Datasets/Transactions/*.csv")
transaction_list=[]

for file in transaction_files: # to add all monthly dataset as one
    df=pd.read_csv(file)
    transaction_list.append(df)

transactions=pd.concat(transaction_list,ignore_index=True)

transactions.shape
```

```python
card_files = glob.glob("../Datasets/Card Status/*.csv")
card_list=[]

for file in card_files: # to add all monthly dataset as one
    df=pd.read_csv(file)
    card_list.append(df)

cards=pd.concat(card_list,ignore_index=True)

cards.shape
```

```python
locations=pd.read_csv("../Datasets/FPS Location/shop-status-details_as_on_3_July_2025.csv")

locations.shape
```

```python
master_df = transactions.merge(
    cards,
    on=['shopNo','distCode','month','year'], # we don't merge on office code, as there are multiple rows (shopno) within one office code
    # also we don't merge on any names as there might be seperate names for same name or place (like, Chennai, CHENNAI or Tarun TarUN) 
    how='left'
)
# we merge the transaction and cards first so mergeing with location would be easier

master_df = master_df.merge(
    locations,
    on=['shopNo','distCode'],
    how='left'
)
```

## Feature Engineering

```python
# to measure number of transactions per ration card

master_df['utilization_ratio'] = (master_df['noOfTrans']/master_df['totalRcs'])

master_df['utilization_ratio'].head(10)
```

```python
# to measure number of transactions from people using other shops 

master_df['portability_ratio'] = (master_df['otherShopTransCnt']/master_df['noOfTrans'])

master_df['portability_ratio'].head(10)
```

```python
# to measure total quantity of major commodities (rice, what, sugar) distribution by a shop

master_df['total_commodity'] = (master_df['riceFsc'] + master_df['wheat'] + master_df['sugar'])

master_df['total_commodity'].head(10)
```

```python
# to measure average commodity distribution per ration card

master_df['commodity_intensity'] = (master_df['total_commodity']/master_df['totalRcs'])

master_df['commodity_intensity'].head(10)
```

```python
master_df['shopNo'].nunique() # this is to chech unique number of shops for shop volatility feature
```

```python
# this is to measure how much transaction volume fluctuates over time for a shop

shop_volatility = (master_df.groupby('shopNo')['noOfTrans'].std().reset_index())
# to group all rows of same shops

shop_volatility.columns = ['shopNo', 'transaction_volatility'] 
# this is just to rename noOfTrans to transaction_volatility as now that columns consists of standard deviation values and not number of transactions

master_df = master_df.merge(shop_volatility, on='shopNo', how='left') 
# we merge actual master dataset and the new shop volatility dataframe

master_df['transaction_volatility'].head(10)
```

```python
# to measure average commodity quantity distributed per transaction

master_df['commodity_per_transaction'] = (master_df['total_commodity']/master_df['noOfTrans'])

master_df['commodity_per_transaction'].head(10)
```

```python
# to measure average number of beneficiary units associated with each ration card

master_df['units_per_card'] = (master_df['totalUnits']/master_df['totalRcs'])

master_df['units_per_card'].head(10)
```

## Feature Scaling & PCA

```python
# we drop these columns as they hinder the clustering based on shop behaviour

drop_cols = [
    'distCode', # because codes doesn't show true similarities of shop behaviour
    'distName',
    'officeCode', # because codes doesn't show true similarities of shop behaviour
    'shopNo', # because shopno doesn't show true similarities of shop behaviour
    'month', # month or year doesn't show shop behaviour
    'year', # we don't want to cluster based on year records like 2023, 2024, 2025
    'rgdal',
    'units',
    'gasCylinders' # rgdal, units, and gasCylinders are mostly the same for all shops and doesn't show behavioural differences
]

cluster_df = master_df.drop(columns=drop_cols)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled_data = scaler.fit_transform(cluster_df)
scaled_df = pd.DataFrame(scaled_data, columns=cluster_df.columns)

scaled_df.head()
```

```python
from sklearn.decomposition import PCA

pca = PCA() # this pca is to check which components/features contribute the most for the dataset
pca.fit(scaled_df)
```

```python
import numpy as np 

cumulative_variance = np.cumsum(pca.explained_variance_ratio_) 
# explained_variance_ratio_ to explain how much of data contribution each principal component does, 
# and then cumsum to see the overall PC contributions, so that we can eliminate the repetitive and unc=ecessarily involved columns

plt.figure(figsize=(8,5))
plt.plot(range(1, len(cumulative_variance)+1), cumulative_variance, marker='o')
plt.xlabel("Number of Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("PCA Explained Variance")
plt.grid(True)
plt.show()
```

This graph shows that most of the data contribution is done by 12 to 15 components, while the remaining 20 features add very little new information.
Hence we choose to use the 12 main components contributing about 95% of the data's information.

```python
from sklearn.decomposition import PCA

pca_2 = PCA(n_components=12) # this is a new PCA for 12 components only, as it already contributes 95% of data
pca_data = pca_2.fit_transform(scaled_df)
pca_df = pd.DataFrame(pca_data, columns=[f'PC{i}' for i in range(1, 13)])

pca_df.head()
```

## K-Means Clustering

```python
# since we don't know how many K we should have for cluster, we can visualise it and check with all K possibilities

from sklearn.cluster import KMeans

wcss = [] # WCSS is Within CLuster Sum of Squares, sum of squares of info within the cluster according to k
# hence we do this to check if each point is closer to or far from the cluster centre according to the k
# if the WCSS is small the clustering is good, otherwise the clustering is bad

for k in range(1, 11): # we check with each number of K to see which one accurately separates behaviours of shop
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    kmeans.fit(pca_df)
    wcss.append(kmeans.inertia_) # inertia_ is just to return the wcss value of the clustering
```

```python
# this visualisation is mainly to check how many clusters we can form with lesser WCSS value for each cluster and its points

plt.figure(figsize=(8,5))
plt.plot(range(1,11), wcss, marker='o')
plt.xlabel("Number of Clusters (k)")
plt.ylabel("WCSS")
plt.title("Optimal Number of Clusters")
plt.grid(True)
plt.show()
```

From k = 4 onwards, the decrease in WCSS becomes much more gradual compared to the earlier values of k. This indicates that adding more clusters after k = 4 provides only marginal improvements in cluster compactness. Therefore, k = 4 is the optimal number of clusters since it represents the point where the rate of improvement starts reducing.

```python
kmeans = KMeans(
    n_clusters=4, # so we create the k means clustering with k=4
    random_state=42,
    n_init=10
)
clusters = kmeans.fit_predict(pca_df)

master_df['cluster'] = clusters # we add this as a column in master_df to adjacently see which rows are part of which cluster

master_df['cluster'].value_counts()
```

```python
cluster_profile = (
    master_df.groupby('cluster')[
        [
            'noOfTrans',
            'totalRcs',
            'otherShopTransCnt',
            'utilization_ratio',
            'portability_ratio',
            'commodity_intensity',
            'transaction_volatility'
        ]
    ].mean()
)

cluster_profile.T.round(2)
```

```python
plt.figure(figsize=(10,6))
plt.scatter(
    pca_df['PC1'],
    pca_df['PC2'],
    c=master_df['cluster'],
    s=5
)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("K-Means Clusters (PCA Projection)")
plt.colorbar(label='Cluster')
plt.show()
```

```python
cluster_profile.plot(kind='bar',figsize=(12,6))

plt.title("Cluster Profile Comparison")
plt.xlabel("Clusters")
plt.ylabel("Average Value")
plt.xticks(rotation=0)
plt.show()
```

```python
# We perform DBSCAN to check which values/rows (shops) are noise and doesn't fit into any clusters
# hence, removing such noise values makes the clustering more accurate

from sklearn.cluster import DBSCAN

dbscan = DBSCAN(eps=1.0, min_samples=10)
dbscan_labels = dbscan.fit_predict(pca_df)
```

```python
pd.Series(dbscan_labels).value_counts()
```

```python
master_df['dbscan_cluster'] = dbscan_labels

master_df['dbscan_cluster'].head()
```
